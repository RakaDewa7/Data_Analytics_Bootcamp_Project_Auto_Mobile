## RFM Analysis (Recency, Frequency, Monetary)

```python
df['ORDERDATE'] = pd.to_datetime(df['ORDERDATE'], format='%d/%m/%Y')

# 3. Tentukan tanggal referensi (1 hari setelah tanggal transaksi paling akhir)
reference_date = df['ORDERDATE'].max() + pd.Timedelta(days=1)

# 4. Agregasi RFM berdasarkan nama pelanggan (CUSTOMERNAME)
rfm = df.groupby('CUSTOMERNAME').agg({
    'ORDERDATE': lambda x: (reference_date - x.max()).days,  # Recency
    'ORDERNUMBER': 'nunique',                                # Frequency
    'SALES': 'sum'                                           # Monetary
}).reset_index()

# Rename kolom agar jelas
rfm.columns = ['CustomerName', 'Recency', 'Frequency', 'Monetary']

# 5. Hitung Skor RFM (skala 1-5 menggunakan qcut/rank)
rfm['R_Score'] = pd.qcut(rfm['Recency'].rank(method='first'), 5, labels=[5, 4, 3, 2, 1]).astype(int)
rfm['F_Score'] = pd.qcut(rfm['Frequency'].rank(method='first'), 5, labels=[1, 2, 3, 4, 5]).astype(int)
rfm['M_Score'] = pd.qcut(rfm['Monetary'].rank(method='first'), 5, labels=[1, 2, 3, 4, 5]).astype(int)

# Gabungkan skor menjadi gabungan string (contoh: '555', '422')
rfm['RFM_Score'] = rfm['R_Score'].astype(str) + rfm['F_Score'].astype(str) + rfm['M_Score'].astype(str)

# 6. Buat aturan segmentasi pelanggan
def assign_segment(row):
    r, f = row['R_Score'], row['F_Score']
    if r >= 4 and f >= 4:
        return 'Champions'
    elif r >= 3 and f >= 3:
        return 'Loyal Customers'
    elif r >= 3 and f <= 2:
        return 'Potential Loyalists'
    elif r <= 2 and f >= 3:
        return 'At Risk'
    else:
        return 'Hibernating Customers'

rfm['Segment'] = rfm.apply(assign_segment, axis=1)
rfm
```

```python
# Menampilkan jumlah pelanggan di setiap segmen
print("\nJumlah Pelanggan per Segmen:")
print(rfm['Segment'].value_counts())
```
